---

В данный момент у вас должен возникнуть вопрос: если Numpy массивы - это "примерно"
```cpp
int narr[2][7];
```
то как в данном формате может работать простая индексация без создания копий?

Люди, знакомые с индексацией многомерных массивов в C могли встречать подобные конструкции:
```cpp
for (int i = 0; i < SHAPE0; i++) {
    for (int j = 0; j < SHAPE1; j++) {
        elem = narr[i][j];

        // the above 2d indexing is actually just syntax sugar for
        elem = *(
            narr
            + i * sizeof(int) * SHAPE1
            + j * sizeof(int)
        );
    }
}
```

В `numpy` значения, на которые умножаются индексы при "обходе" массива (`sizeof(int) * SHAPE1` и `sizeof(int)` в примере выше) называются `strides`. В отличие от примера на C, в `numpy` страйды не известны в момент компиляции, а хранятся как метаданные в `ndarray` объектах.

То есть на самом деле в `numpy` массивы не
```cpp
int narr[2][7];
```

А что-то ближе к
```cpp
struct ndarray {
    void *data;
    // some extra metadata:
    // - dtype
    // - ndim
    // - shape
    // - strides
    // - flags
    // - etc
} narr;
```

````
narr.dtype.itemsize # sizeof(dtype)
````

````
# (  sizeof(int)     *      SHAPE1,       sizeof(int)  )
(narr.dtype.itemsize * narr.shape[1], narr.dtype.itemsize)
````

````
narr.strides
````

````
narr.shape
````

Подобное "динамическое" представление массивов позволяет производить многие "простые" операции над массивами просто изменяя значения метаданных `ndim`, `shape`, `strides` или сдвигая указатель на "начало" массива.

В частности:

- любая **простая** индексация (в том числе индексами, слайсами или их комбинациями)
- перемещение осей (`transpose`)
- некоторые "простые" операции изменения формы (`[None]`, `.squeeze()`, `reshape` *см. следующий пункт)
- повторение вдоль осей с `shape == 1` (`broadcasting` **(!)**, это делается устанавливая `stride=0`)

могут быть выполнены изменяя только метаданные массива, не трогая сами данные лежащие по указателю `data`.

Из "простых" операций, пожалуй единственное исключение, при котором **не всегда** создается `view` - это `reshape`.

По умолчанию `reshape` старается не делать копии, если это возможно. Однако, если изменение формы не возможно достичь просто изменением метаданных `shape` и `strides`, то `reshape` делает копию:

````
A = np.arange(24).reshape(6, 4)
A
````

````
# Transpose always returns a view
B = A.T # view
B[0, 0] = 9
B
````

````
A
````

````
# Reshape **sometimes** returns a view
C = A.reshape(-1) # view
C[0] = 99
C
````

````
A
````

````
# But **sometimes** it returns a copy
D = B.reshape(-1) # implicit copy (!)
D[0] = 999
D
````

````
C
````

````
A
````

````
A.reshape(-1, copy=False)
````

````
B.reshape(-1, copy=False)
````

Упражнение на дом:
- Посмотрите, как `A.reshape(-1, copy=False)` изменяет `shape` и `strides` различных массивов.<br/>&nbsp;
- Подумайте, почему `B.reshape(-1, copy=False)` в данном примере не возможно выполнить без копии, изменяя только `shape` и `strides`.

**Сложную** индексацию (массивом индексов или булевой маской) невозможно свести к операциям только над `ndim`, `shape`, `strides` и указателем на начало.

Как следсвтие, сложная индексация **всегда** возвращает копию массива:

````
A = np.arange(24).reshape(2, 3, 4)
A
````

````
b = A[A > 5]
b += 1000
````

````
A, b
````

````
c = A[:, :, [0]]
c += 2000
````

````
A, c
````

Но с другой стороны in-place операции все ещё работает при сложной индексации:

````
A[A > 5] -= 1000
A
````

````
A[:, :, [0, 1]] -= 2000
A
````
